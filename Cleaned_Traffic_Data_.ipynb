```python
import pandas as pd
```

```python
df = pd.read_csv('/content/amazon-web-traffic-dataset.csv')
```

# Fix Date

```python
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')
```

# Quick overview

```python
print(df.info())
print(df.describe())
```

# 1. Traffic Sources Analysis

```python
traffic_sources = df.groupby('Source').agg(
    total_users=('New Users', 'sum'),
    total_conversions=('Conversions', 'sum'),
    avg_bounce_rate=('Bounce Rate', 'mean')
).reset_index()
traffic_sources['conversion_rate'] = (traffic_sources['total_conversions'] / traffic_sources['total_users'] * 100).round(2)
traffic_sources = traffic_sources.sort_values('total_users', ascending=False)
print(traffic_sources)
```

# 2. Engagement metrics

```python
engagement = df.agg(
    avg_session_duration=('Avg Session Duration', 'mean'),
    avg_pageviews=('Page Views', 'mean'),
    avg_bounce_rate=('Bounce Rate', 'mean'),
    avg_time_home=('Average time on home page (min)', 'mean')
)
print(engagement)
```

# 3. Device Breakdown

```python
device_stats = df.groupby('Device Category').agg(
    total_users=('New Users', 'sum'),
    avg_duration=('Avg Session Duration', 'mean'),
    avg_bounce=('Bounce Rate', 'mean'),
    conversions=('Conversions', 'sum')
).reset_index()
print(device_stats)
```

# 4. Page Performance

```python
page_stats = df.groupby('Page Path').agg(
    total_pageviews=('Page Views', 'sum'),
    total_unique_views=('Unique Page Views', 'sum'),
    avg_bounce=('Bounce Rate', 'mean')
).reset_index().sort_values('total_pageviews', ascending=False)
print(page_stats)
```

# 5. Day of week pattern

```python
day_stats = df.groupby('Day').agg(
    total_users=('New Users', 'sum'),
    conversions=('Conversions', 'sum'),
    avg_bounce=('Bounce Rate', 'mean')
).reset_index()
print(day_stats)
```

# 6. Country Performance

```python
country_stats = df.groupby('Country').agg(
    total_users=('New Users', 'sum'),
    conversions=('Conversions', 'sum'),
    avg_bounce=('Bounce Rate', 'mean')
).reset_index().sort_values('total_users', ascending=False)
print(country_stats)
```

# 7. Daily Trend

```python
daily_trend = df.groupby('Date').agg(
    total_pageviews=('Page Views', 'sum'),
    total_conversions=('Conversions', 'sum')
).reset_index()
print(daily_trend.head())
```
